# Langkah 1: Impor library yang diperlukan


```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from sklearn.preprocessing import LabelEncoder
```

# Langkah 2  : Ambil data

- simpan data kedalam variable = df_NPM
- simpan data varibale bebas kedalam = X_NPM
- simpan data variable terikat/target kedalam = y_NPM

```python
df_714220016 = pd.read_csv('Social_Network_Ads.csv')
```

```python
df_714220016.head()
```

```python
df_714220016.info()
```

# Langkah 3 : Membagi data set menjadi data train dan testing

```python
#Memisahkan kumpulan data ke dalam kumpulan Pelatihan dan kumpulan Pengujian
X_714220016 = df_714220016[['Age', 'EstimatedSalary']]

y_714220016 = df_714220016['Purchased']

print(X_714220016.head())

print("\nTarget (y_714220016):")
print(y_714220016.head())
```

```python
X_train_714220016, X_test_714220016, y_train_714220016, y_test_714220016 = train_test_split(X_714220016, y_714220016, test_size=0.25, random_state=0)

X_test_714220016 = X_test_714220016.to_numpy()
print("\nTesting set (X_test_714220016):")
print(X_test_714220016)
```

# Langkah 4: Standarisasi (normalisasi) skala data untuk mempersiapkan algoritma KNN

**Feature Scaling adalah suatu cara untuk membuat numerical data pada dataset memiliki rentang nilai (scale) yang sama**
- simpan fungsi StandardScaler kedalam variable = sc_NPM

```python
sc_714220016 = StandardScaler()
X_train_scaled_714220016 = sc_714220016.fit_transform(X_train_714220016)
X_test_scaled_714220016 = sc_714220016.transform(X_test_714220016)
```

Menampikan data test yang sudah dilakukan fungsi perubahan skalar

```python
print("\nScaled X_test_scaled_714220016 head:")
print(X_test_scaled_714220016)
np.set_printoptions(threshold=np.inf)
X_test_scaled_714220016
```

## Langkah 5 : Memilih Nilai K

- simpan hasil kesalahan dalam variable = ER_NPM

```python
ER_714220016 = []
for i in range(1, 40):
    knn_714220016 = KNeighborsClassifier(n_neighbors=i)
    knn_714220016.fit(X_train_scaled_714220016, y_train_714220016)
    y_pred_714220016 = knn_714220016.predict(X_test_scaled_714220016)
    error = 1 - accuracy_score(y_test_714220016, y_pred_714220016)
    ER_714220016.append(error)
```

### Visualisasi Metode Elbow

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
plt.plot(range(1, 40), ER_714220016, marker='o', linestyle='-', color='b')
plt.title('KNN Accuracy vs Number of Neighbors')
plt.xlabel('Number of Neighbors (k)')
plt.ylabel('Error Rate')
plt.xticks(range(1, 40, 2))
plt.grid()
plt.show()
```

# Langkah 6: Buat dan Latih Model

## K = 1

- Kita akan gunakan K = 1
- simpan fungsi KNeighborsClassifier ke dalam variable = knn_NPM

```python
from sklearn.neighbors import KNeighborsClassifier
knn_714220016 = KNeighborsClassifier(n_neighbors=1)
knn_714220016.fit(X_train_scaled_714220016, y_train_714220016)
```

### Buat Prediksi  dari K = 1

- simpan hasil prediksi pada variable = pred_K1_NPM

```python
pred_k1_714220016 = knn_714220016.predict(X_test_scaled_714220016)
```

```python
print("\nPredictions with k=1:")
np.array(pred_k1_714220016)
```

### Melihat hasil prediksi dari data input dalam dataset

```python
for fitur, prediksi in zip(X_test_scaled_714220016, pred_k1_714220016):
    print(f"Fitur: {fitur}, Prediksi: {prediksi}")
```

### Evaluasi hasil model dengan K=1

```python
from sklearn.metrics import classification_report,confusion_matrix
```

```python
confusion_matrix(y_test_714220016, pred_k1_714220016)
```
