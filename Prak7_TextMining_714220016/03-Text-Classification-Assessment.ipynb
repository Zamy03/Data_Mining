```python
import pandas as pd
import numpy as np
df_714220016 = pd.read_csv('moviereviews.tsv', sep='\t')
df_714220016.head()




```

### Task #2: Check for missing values:

```python
# Check for NaN values:
missing_values = df_714220016.isnull().sum()
print("jumlah missing values")
print(missing_values)
```

```python
# Check for whitespace strings (it's OK if there aren't any!):
print("jumlah whitespace strings")
whitespace_strings_count = {}
for column in df_714220016.select_dtypes(include=[object]).columns:
    is_whitespace_or_empty_after_strip = (df_714220016[column].astype(str).str.strip() == '') & (~df_714220016[column].isnull())
    count = is_whitespace_or_empty_after_strip.sum()
    if count > 0:
        whitespace_strings_count[column] = count

if whitespace_strings_count:
    print("Jumlah 'whitespace strings' (string yang hanya berisi spasi/kosong setelah strip) per kolom:")
    for col, count in whitespace_strings_count.items():
        print(f"- Kolom '{col}': {count} entri")
else:
    print("Tidak ditemukan 'whitespace strings' di kolom bertipe string.")
```

### Task #3: Remove NaN values:

```python
# hapus NaN
print("shape sebelum menghapus NaN", df_714220016.shape)
initial_rows = df_714220016.shape[0]

print("\nMemulai proses menghapus baris yang mengandung NaN...")
df_714220016_after_nan_removal = df_714220016.dropna(how='any')
print("Shape DataFrame setelah penghapusan NaN:", df_714220016_after_nan_removal.shape)
rows_removed_nan = initial_rows - df_714220016_after_nan_removal.shape[0]
print(f"Jumlah baris yang dihapus karena NaN: {rows_removed_nan}")

# hapus whitespace strings
print("\nMemulai proses menghapus baris yang mengandung hanya whitespace strings...")
string_columns = df_714220016_after_nan_removal.select_dtypes(include='object').columns

whitespace_mask = pd.Series([False] * len(df_714220016_after_nan_removal), index=df_714220016_after_nan_removal.index)
for col in string_columns:
    col_is_whitespace = df_714220016_after_nan_removal[col].astype(str).str.strip() == ''
    whitespace_mask = whitespace_mask | col_is_whitespace # Combine masks with OR operation

df_714220016_clean_final = df_714220016_after_nan_removal[~whitespace_mask]
print("Shape DataFrame setelah penghapusan whitespace strings:", df_714220016_clean_final.shape)
rows_removed_whitespace = df_714220016_after_nan_removal.shape[0] - df_714220016_clean_final.shape[0]
print(f"Jumlah baris yang dihapus karena whitespace strings: {rows_removed_whitespace}")

total_rows_removed = rows_removed_nan + rows_removed_whitespace
print(f"Total jumlah baris yang dihapus: {total_rows_removed}")

print("\n5 baris pertama DataFrame setelah pembersihan:")
print(df_714220016_clean_final.head())

print("\nJumlah missing values setelah pembersihan:")
print(df_714220016_clean_final.isnull().sum())

print("\nPengecekan ulang whitespace strings (seharusnya tidak ada):")
whitespace_check_final = {}
for column in df_714220016_clean_final.select_dtypes(include=[object]).columns:
    is_whitespace_or_empty_after_strip = (df_714220016_clean_final[column].astype(str).str.strip() == '') & (~df_714220016_clean_final[column].isnull())
    count = is_whitespace_or_empty_after_strip.sum()
    if count > 0:
        whitespace_check_final[column] = count
if whitespace_check_final:
    print("Masih ada 'whitespace strings' di kolom berikut:")
    for col, count in whitespace_check_final.items():
        print(f"- Kolom '{col}': {count} entri")
else:
    print("Tidak ditemukan 'whitespace strings' di DataFrame setelah pembersihan.")

```

### Task #4: Take a quick look at the `label` column:

### Task #5: Split the data into train & test sets:
You may use whatever settings you like. To compare your results to the solution notebook, use `test_size=0.33, random_state=42`.

```python
from sklearn.model_selection import train_test_split

#split data
print("\nMemisahkan data menjadi fitur (x) dan label (y)...")
x_714220016 = df_714220016_clean_final['review']
y_714220016 = df_714220016_clean_final['label']

x_714220016_train, x_714220016_test, y_714220016_train, y_714220016_test = train_test_split(x_714220016, y_714220016, test_size=0.33, random_state=42)


print("Data berhasil dibagi.")
print(f"Shape X_train: {x_714220016_train.shape}")
print(f"Shape X_test: {x_714220016_test.shape}")
print(f"Shape y_train: {y_714220016_train.shape}")
print(f"Shape y_test: {y_714220016_test.shape}")
```

### Task #6: Build a pipeline to vectorize the date, then train and fit a model
You may use whatever model you like. To compare your results to the solution notebook, use `LinearSVC`.
---
**rule**:
**npm_model**

```python
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.svm import LinearSVC
from sklearn.pipeline import Pipeline
from sklearn.metrics import classification_report, accuracy_score

# Membuat pipeline untuk TF-IDF dan model SVM
pipeline_714220016 = Pipeline([
    ('tfidf', TfidfVectorizer()),
    ('svm', LinearSVC())
])

# Melatih model
pipeline_714220016.fit(x_714220016_train, y_714220016_train)
```

### Task #7: Run predictions and analyze the results

---
**rule: npm_pred**


```python
# Form a prediction set
print("\nMembuat prediksi pada data uji...")
predictions_714220016 = pipeline_714220016.predict(x_714220016_test)
print("Prediksi berhasil dibuat.")
```

```python
# Report the confusion matrix
from sklearn.metrics import confusion_matrix, classification_report, accuracy_score # Import confusion_matrix
print("\nConfusion Matrix:")
cm = confusion_matrix(y_714220016_test, predictions_714220016)
print(cm)
```

```python
# Print a classification report
print("\nClassification Report:")
print(classification_report(y_714220016_test, predictions_714220016))
```

```python
# Print the overall accuracy
print(f"Akurasi Model: {accuracy_score(y_714220016_test, predictions_714220016):.4f}")
```

## Task #8 : Apply Stopword

```python
#Membuat Pipeline dengan Stopword Removal di TfidfVectorizer
pipeline_714220016_stopwords = Pipeline([
    ('tfidf', TfidfVectorizer(stop_words='english')),
    ('svm', LinearSVC())
])

# Melatih model dengan stopword removal
pipeline_714220016_stopwords.fit(x_714220016_train, y_714220016_train)
```

```python
#prediksi dengan model yang menggunakan stopword removal
predictions_714220016_stopwords = pipeline_714220016_stopwords.predict(x_714220016_test)
# Report the confusion matrix untuk model dengan stopword removal
print("\nConfusion Matrix dengan Stopword Removal:")
cm_stopwords = confusion_matrix(y_714220016_test, predictions_714220016_stopwords)
print(cm_stopwords)
# Print a classification report untuk model dengan stopword removal
print("\nClassification Report dengan Stopword Removal:")
print(classification_report(y_714220016_test, predictions_714220016_stopwords))
# Print the overall accuracy untuk model dengan stopword removal
print(f"Akurasi Model dengan Stopword Removal: {accuracy_score(y_714220016_test, predictions_714220016_stopwords):.4f}")
```

## Task #9 : Compare to other Algortihm

```python
#Knn Classifier
from sklearn.neighbors import KNeighborsClassifier

# Membuat pipeline untuk TF-IDF dan model KNN
pipeline_714220016_knn = Pipeline([
    ('tfidf', TfidfVectorizer(stop_words='english')),
    ('knn', KNeighborsClassifier(n_neighbors=5))
])
# Melatih model KNN
pipeline_714220016_knn.fit(x_714220016_train, y_714220016_train)

# Form a prediction set dengan model KNN
predictions_714220016_knn = pipeline_714220016_knn.predict(x_714220016_test)

# Report the confusion matrix untuk model KNN
print("\nConfusion Matrix untuk KNN:")
cm_knn = confusion_matrix(y_714220016_test, predictions_714220016_knn)
print(cm_knn)

# Print a classification report untuk model KNN
print("\nClassification Report untuk KNN:")
print(classification_report(y_714220016_test, predictions_714220016_knn))

# Print the overall accuracy untuk model KNN
print(f"Akurasi Model KNN: {accuracy_score(y_714220016_test, predictions_714220016_knn):.4f}")
```

```python
#Naive Bayes
from sklearn.naive_bayes import MultinomialNB
# Membuat pipeline untuk TF-IDF dan model Naive Bayes
pipeline_714220016_nb = Pipeline([
    ('tfidf', TfidfVectorizer(stop_words='english')),
    ('nb', MultinomialNB())
])

# Melatih model Naive Bayes
pipeline_714220016_nb.fit(x_714220016_train, y_714220016_train)

# Form a prediction set dengan model Naive Bayes
predictions_714220016_nb = pipeline_714220016_nb.predict(x_714220016_test)

# Report the confusion matrix untuk model Naive Bayes
print("\nConfusion Matrix untuk Naive Bayes:")
cm_nb = confusion_matrix(y_714220016_test, predictions_714220016_nb)
print(cm_nb)

# Print a classification report untuk model Naive Bayes
print("\nClassification Report untuk Naive Bayes:")
print(classification_report(y_714220016_test, predictions_714220016_nb))

# Print the overall accuracy untuk model Naive Bayes
print(f"Akurasi Model Naive Bayes: {accuracy_score(y_714220016_test, predictions_714220016_nb):.4f}")
```

```python
#logistic regression
from sklearn.linear_model import LogisticRegression

# Membuat pipeline untuk TF-IDF dan model Logistic Regression
pipeline_714220016_lr = Pipeline([
    ('tfidf', TfidfVectorizer(stop_words='english')),
    ('lr', LogisticRegression(max_iter=1000))
])

# Melatih model Logistic Regression
pipeline_714220016_lr.fit(x_714220016_train, y_714220016_train)

# Form a prediction set dengan model Logistic Regression
predictions_714220016_lr = pipeline_714220016_lr.predict(x_714220016_test)

# Report the confusion matrix untuk model Logistic Regression
print("\nConfusion Matrix untuk Logistic Regression:")
cm_lr = confusion_matrix(y_714220016_test, predictions_714220016_lr)
print(cm_lr)

# Print a classification report untuk model Logistic Regression
print("\nClassification Report untuk Logistic Regression:")
print(classification_report(y_714220016_test, predictions_714220016_lr))

# Print the overall accuracy untuk model Logistic Regression
print(f"Akurasi Model Logistic Regression: {accuracy_score(y_714220016_test, predictions_714220016_lr):.4f}")
```
