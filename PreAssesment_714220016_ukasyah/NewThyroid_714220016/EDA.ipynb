```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("new-thyroid.csv")
df.info()
```

```python
stat_summary = df.describe().loc[['mean', 'std', 'min', 'max']]
stat_summary
```

```python
plt.figure(figsize=(14, 6))

# Histogram dan boxplot untuk fitur '10.1'
plt.subplot(2, 2, 1)
sns.histplot(df['10.1'], kde=True, bins=20)
plt.title("Histogram - Fitur 10.1")

plt.subplot(2, 2, 2)
sns.boxplot(y=df['10.1'])
plt.title("Boxplot - Fitur 10.1")

# Histogram dan boxplot untuk fitur '0.9'
plt.subplot(2, 2, 3)
sns.histplot(df['0.9'], kde=True, bins=20)
plt.title("Histogram - Fitur 0.9")

plt.subplot(2, 2, 4)
sns.boxplot(y=df['0.9'])
plt.title("Boxplot - Fitur 0.9")

plt.tight_layout()
plt.show()
```

```python
# Korelasi matriks
correlation_matrix = df.corr()

# Heatmap
plt.figure(figsize=(10, 8))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm')
plt.title("Heatmap Korelasi Fitur Numerik")
plt.show()
```

```python
sns.countplot(x=df['1'])
plt.title("Distribusi Label / Kelas")
plt.xlabel("Kelas")
plt.ylabel("Jumlah")
plt.show()
```
