```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('ruspini.csv')
```

```python
df.info()
```

```python
df[['X', 'Y']].describe().T[['min', 'max', 'mean', 'std']]
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.set(style="whitegrid")
fig, axes = plt.subplots(2, 2, figsize=(12, 10))

# Histogram & Boxplot fitur X
sns.histplot(df['X'], kde=True, ax=axes[0, 0], color='skyblue')
axes[0, 0].set_title('Histogram Fitur X')
sns.boxplot(x=df['X'], ax=axes[1, 0], color='lightgreen')
axes[1, 0].set_title('Boxplot Fitur X')

# Histogram & Boxplot fitur Y
sns.histplot(df['Y'], kde=True, ax=axes[0, 1], color='salmon')
axes[0, 1].set_title('Histogram Fitur Y')
sns.boxplot(x=df['Y'], ax=axes[1, 1], color='lightyellow')
axes[1, 1].set_title('Boxplot Fitur Y')

plt.tight_layout()
plt.show()
```

```python
# Heatmap korelasi
plt.figure(figsize=(6, 4))
corr_matrix = df.drop(columns=['#']).corr()
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title("Heatmap Korelasi Antar Fitur")
plt.show()

```

```python
# Visualisasi sebaran CLASS
plt.figure(figsize=(6, 4))
sns.countplot(x='CLASS', data=df, palette='pastel')
plt.title('Sebaran Label CLASS')
plt.xlabel('Kelas')
plt.ylabel('Jumlah')
plt.show()
```
